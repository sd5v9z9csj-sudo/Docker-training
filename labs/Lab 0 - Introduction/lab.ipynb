# Welcome to the AI Labs

These labs are served through JupyterHub, a Jupyter environment prepared for students and available directly in your web browser. You can open notebooks, read instructions, run selected cells when required, and inspect outputs without installing anything locally. Some labs are text-only and do not require you to run or edit anything.

When you run code from a notebook, it executes inside a Kubernetes pod created for your Jupyter session. Any interaction with the Kubernetes API from the notebook uses the Kubernetes service account assigned to that pod, so available actions depend on the permissions granted to that service account.

A notebook is made of cells. Each cell can contain either code or Markdown text. In these labs, most cells are provided as read-only instructions or prepared code; only cells with explicit placeholders should be edited.

## Editing Lab Cells

These labs are not meant to be freely modified. Most cells are read-only and should be treated as instructions, explanations, or provided code.

Only edit cells that clearly ask for your input. Required input is shown with an obvious placeholder, usually in uppercase, such as `DOCKER_USERNAME`, `YOUR_PROJECT_ID`, or `FILL_ME_IN`. Replace only the placeholder value and leave the rest of the cell unchanged.

If a cell does not contain a clear placeholder or explicit instruction to edit it, do not change it.

## Using Jupyter Cells

There are two main cell types:

- **Markdown cells** are used for formatted text, explanations, headings, links, and lists.
- **Code cells** can run commands or programs and show their output directly below the cell. Depending on the lab, they may run Bash commands, download files, execute `kubectl`, or run Python programs.

Useful shortcuts:

- `Shift + Enter`: run only the selected cell and move to the next one.
- `Ctrl + Enter`: run only the selected cell and stay on it.
- `Alt + Enter`: run only the selected cell and create a new cell below it. Because it changes the notebook structure, avoid it unless you specifically need a new temporary cell.
- `Esc`: leave text-editing mode and enter command mode.
- `Enter`: edit the selected cell.

If a Markdown cell is showing raw Markdown because it was double-clicked by mistake, re-render only that selected cell with `Shift + Enter` or `Ctrl + Enter`. Do not use the toolbar **Run** button or any **Run All** option, because these can start executing and rendering many cells, including code cells.

## Practical Recommendations

- Do not click global **Run** or **Run All** controls. They may execute the whole notebook, including all code cells.
- When a lab asks you to run code, run only the specific selected cell or cells described by the instructions.
- Some labs may include a reset cell near the beginning. A reset cell restores the lab infrastructure to the expected starting state. Run it only when the lab instructions tell you to reset the environment.
- If results look inconsistent, restart the kernel and rerun only the specific cells needed for the current task.
- Check the symbol next to a code cell: `[*]` means the cell is still running, while a number such as `[3]` means it has finished. The cell may also briefly flash when execution completes.
- Do not rely only on browser refresh. If the notebook becomes unresponsive, restart the kernel first.
