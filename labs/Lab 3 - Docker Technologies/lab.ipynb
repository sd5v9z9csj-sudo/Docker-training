## Start the lab

Run the lab environment from the terminal.

```bash
cd ~/lab/docker-lab
./start-labspace.sh
```

Wait until all components report a running state. Open [http://localhost:3030](http://localhost:3030) to begin the `Docker` lab.

> Tip: If the page does not load immediately, check whether `docker compose` is still downloading images or rebuilding content.
